```python
#Imported all necessary libraries for this random forest dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics
```

```python
from google.colab import files #imported the dataset into the google colab notebook
uploaded=files.upload()
df=pd.read_csv('nyc_housing_base.csv')#This file consisted of house details in NYC
df.head()
```

```python
df.isnull().sum() #Checking if null values are present in the data set
```

```python
df= df.dropna(subset=['zip_code']) #Dropped the null values in this columns
df
```

```python
# The rest of the columns will have all of the null values dropped
```

```python
df= df.dropna(subset=['resarea'])
```

```python
df= df.dropna(subset=['comarea'])
```

```python
df= df.dropna(subset=['numfloors'])
```

```python
df= df.dropna(subset=['landuse'])
```

```python
df= df.dropna(subset=['latitude'])
```

```python
df.isnull().sum() #Double checking to see if the null values have been dropped.
```

```python
df.drop('bldgclass', axis=1, inplace=True) #Dropped these columns as they are not needed for the decision tree
df.drop('borough_y', axis=1, inplace=True)
```

```python
df.drop('latitude', axis=1, inplace=True) #These were also dropped, we dropped them to get the best accuracy possible of the tree.
df.drop('longitude', axis=1, inplace=True)
```

```python
#Most of the columns were initially in float form, so we changed them into a integer form
#This will make it easier later when making our prediction
columns_to_convert = ['zip_code', 'yearbuilt', 'lotarea', 'bldgarea', 'resarea', 'comarea', 'unitsres', 'unitstotal', 'numfloors', 'landuse', 'building_age']
df[columns_to_convert] = df[columns_to_convert].astype(int)
```

```python
df
```

```python
#We are splitting the dataset, the X variables will determine the building age
#We made the building age the target variable
x_data=df[['borough_x', 'block', 'lot', 'sale_price', 'zip_code', 'yearbuilt', 'lotarea', 'bldgarea', 'resarea', 'comarea', 'unitsres', 'unitstotal', 'numfloors', 'landuse' ]]
y_data=df['building_age']
```

```python
#This will split the dataset inot a traning and test set
x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.2, random_state=1)
```

```python
# This trains a single decision tree
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(x_train, y_train)

#This will calculate the accuracy with the dt_accuracy variable
dt_accuracy = dt_model.score(x_test, y_test)
#Prints the final accuracy %
print("Decision Tree Test Accuracy:", f"{dt_accuracy * 100:.2f}%")
```

```python
#This will train the random forest with a random feature
rf_model = RandomForestClassifier(n_estimators=14, max_features=2, random_state=42)

rf_model.fit(x_train, y_train)

#This will calculate the accuracy of the forest
#Instead with a rf_accuracy
rf_accuracy = rf_model.score(x_test, y_test)

#prints the final accuracy
print("RF Model Accuracy:", f"{rf_accuracy *100:.2f}", "%")
```

```python
#This will train the random forest
#It prepares the dataset for a precision, recall, and f1 %
rf_model = RandomForestClassifier(n_estimators=14, max_features=2, random_state=42)

rf_model.fit(x_train, y_train)
#This will make the prediction of the testing set
y_pred_rf = rf_model.predict(x_test)

#Calculation the precision, recall, and the F1 score for the random forest
precision_rf = metrics.precision_score(y_test, y_pred_rf, average='weighted')
recall_rf = metrics.recall_score(y_test, y_pred_rf, average='weighted')


f1_rf = metrics.f1_score(y_test, y_pred_rf, average='weighted')
#Prints the final % for all three
print("RF Precision:", precision_rf)
print("RF Recall:", recall_rf)
print("RF F1-Score:", f1_rf)
```

```python
#This will prepare a new entry and a prediction
def predict_houses():
    while True:
        try:
            borough_x = int(input("Enter the Borough Number: "))
            block = int(input("Enter the block number: "))
            lot = int(input("Enter the lot number: "))
            sale_price = int(input("Enter the sale price:  "))
            zip_code = int(input("Enter the zip code: "))
            yearbuilt = int(input("Enter the year built: "))
            lotarea = int(input("Enter the lot area number: "))
            bldgarea = int(input("Enter the building area number: "))
            resarea = int(input("Enter the res area number: "))
            comarea = int(input("Enter the com area number: "))
            unitsres = int(input("Enter the unitres number: "))
            unitstotal = int(input("Enter the total units: "))
            numfloors = int(input("Enter the number of floors: "))
            landuse = int(input("Enter the land use number: "))
#This will validate the values and make sure there aren't any negative values inputted
            if any(val < 0 for val in [borough_x, block, lot, sale_price, zip_code, yearbuilt, lotarea, bldgarea, resarea, comarea, unitsres, unitstotal, numfloors, landuse]):
                print("Error: Please enter a non-negative value for all feautures.")
                continue
#This will start the prediction of the "new" dataset that we have created
            new_house_data = pd.DataFrame([[borough_x, block, lot, sale_price, zip_code, yearbuilt, lotarea, bldgarea, resarea, comarea, unitsres, unitstotal, numfloors, landuse]],
                                          columns= ['borough_x', 'block', 'lot', 'sale_price', 'zip_code', 'yearbuilt', 'lotarea', 'bldgarea', 'resarea', 'comarea', 'unitsres', 'unitstotal', 'numfloors', 'landuse'])
            prediction = rf_model.predict(new_house_data)[0]
#This will determine if the house is under or over 100 years old
            if prediction > 100:
                #It will print a message for either outcome
                print("Prediction: The house is at least 100 years old ")
            else:
                print("Prediction: The house is less than 100 years old:")
#This gives the option to enter new data for a new outcome
            another_house = input("Enter data for another house? (yes/no): ")
            if another_house.lower() != 'yes':
                break
        except ValueError as e: #if there is a a error it will have you restart and enter the correct values
              print("Invalid input. Please enter a valid number.")
              print("Error:", e)

predict_houses() #This will start the prediction

```
