```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.preprocessing import LabelEncoder
```

```python
#read dataset
df=pd.read_csv('credit_risk.csv')
df.head(20)
```

```python
#any null variables?
df.isnull().sum()
```

```python
#drop null data
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df.info()
```

```python
#label encoder for strings
le=LabelEncoder()
df['person_home_ownership']=le.fit_transform(df['person_home_ownership'])
df['loan_intent']=le.fit_transform(df['loan_intent'])
df['cb_person_default_on_file']=le.fit_transform(df['cb_person_default_on_file'])
df.head(20)
```

```python
df.columns
```

```python
#split dataset in features and target variable
x_data=df[['person_age', 'person_income', 'person_home_ownership',
       'person_emp_length', 'loan_intent', 'loan_amnt', 'loan_int_rate', 'loan_percent_income', 'cb_person_default_on_file',
       'cb_person_cred_hist_length']]
y_data=df['loan_status']
```

```python
# Split dataset into training set and test set
X_train, X_test, y_train, y_test = train_test_split(x_data, y_data, test_size=0.3, random_state=1)
```

```python
#train a single Decision tree
dt_model=DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

# Calculate the accuracy and assign it to dt_accuracy
dt_accuracy = dt_model.score(X_test, y_test)

print("Decision Tree Test Accuracy:", f"{dt_accuracy * 100:.2f}%")
```

```python

#train a Random Forest with random feature selection
rf_model = RandomForestClassifier(n_estimators=5, max_features=2, random_state=42)  # max_features=2 means only 2 features are considered at each split


rf_model.fit(X_train, y_train)


#calculate the accuracy and assign it to rf_accuracy
rf_accuracy = rf_model.score(X_test, y_test)

# Now you can use rf_accuracy in the print statements below
print("RF Model Accuracy:", f"{rf_accuracy * 100:.2f}", "%")

```

```python
# Make predictions on the test set
y_pred_rf = rf_model.predict(X_test)

# Calculate precision, recall, and F1-score for Random Forest
precision_rf = metrics.precision_score(y_test, y_pred_rf)
recall_rf = metrics.recall_score(y_test, y_pred_rf)


f1_rf = metrics.f1_score(y_test, y_pred_rf)


print("RF Precision:", precision_rf)
print("RF Recall:", recall_rf)
print("RF F1-Score:", f1_rf)
```

```python
# Prepare a provision for a new data entry and its prediction of an outcome.

def predict_loan():
    while True:
      try:
        person_age=int(input("Enter Age:"))
        person_income=int(input("Enter Income:"))
        person_home_ownership = int(input("Enter Home Ownership(0=Rent, 2=Own, 3=Mortgage, 1=Other): "))
        person_emp_length = float(input("Enter Employment Length"))
        loan_intent = int(input("Enter Loan Intent(1=Educational, 0=Venture, 2=Home, 4=Personal, 3=Medical, 5=Debt):"))
        loan_amnt = int(input("Enter Loan Amount:"))
        loan_int_rate = float(input("Enter Loan Interest Rate float:"))
        loan_percent_income = float(input("Enter Loan % Income float:"))
        cb_person_default_on_file = int(input("Is your account on file? (0=No,1=Yes):"))
        cb_person_cred_hist_length = int(input("Enter Credit History Length:"))

            # Input validation (example: check for non-negative values)
        if any(val < 0 for val in [person_age, person_income, person_home_ownership, person_emp_length, loan_intent, loan_amnt, loan_int_rate, loan_percent_income, cb_person_default_on_file, cb_person_cred_hist_length]):
                print("Error: Please enter non-negative values for all features.")
                continue  #restart the loop

        new_credit_data = pd.DataFrame([[person_age, person_income, person_home_ownership,
       person_emp_length, loan_intent, loan_amnt, loan_int_rate, loan_percent_income, cb_person_default_on_file,
       cb_person_cred_hist_length]], columns=['person_age', 'person_income', 'person_home_ownership', 'person_emp_length', 'loan_intent', 'loan_amnt', 'loan_int_rate', 'loan_percent_income', 'cb_person_default_on_file', 'cb_person_cred_hist_length'])

        prediction =rf_model.predict(new_credit_data)[0]

        if prediction == 1:
                print("Prediction: Approved for Loan.")
        else:
                print("Prediction: Denied Loan.")

        another_loan = input("Enter data for another loan? (yes/no): ")
        if another_loan.lower() != 'yes':
                break #exit loop
      except ValueError:
            print("Invalid input. Please enter valid numerical values.")

predict_loan()

```
