```python
#Importing all necessary libraries for this algorithm
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer
import numpy as np
import pandas as pd
```

```python
#Importing a dataset full of short stories into the google colab notebook
from google.colab import files
uploaded = files.upload()
df=pd.read_csv('validation.csv')
df
```

```python
#checking for any null values in the dataset
df.isnull().sum()
```

```python
#Create variables for the algotithm
#This dataset only contained one column so we had to make new labels for the text
story = df["text"].astype(str)
labels = story.apply(lambda x: 1 if len(x.split()) > 40 else 0)
```

```python
#this tool will turn the words into numbers and help with seperating the labels
vectorizer = CountVectorizer()
#This will fit or learn all of the unique words
#converting all the words into numbers
#x is a bigtable of numbers
x = vectorizer.fit_transform(story)
#importing new libraries in order to make this classification report
#the libraries are necessary to train and test split the data
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, labels, test_size=0.2)

model = MultinomialNB() # creates a blank model that is untrained
model.fit(x_train, y_train) # this will start training the model
#creates a prediction of the testing model
y_pred = model.predict(x_test)
#importing new libraries
#The rest of the code will start making the classification report

from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
print("=== STEP 3: Words converted to numbers ===")
print(f"Number of unique words found: {len(vectorizer.get_feature_names_out())}")
print(f"All unique words: {vectorizer.get_feature_names_out()}")
print()
#This model seems to be perfect as it is 1.00
```

```python
#we are creating a new variable and identifying if there are fake stories in the dataset
new_story = [
    "Once upon a time there was a brave knight",
    "Hi can you send the file",
]
#converting the new variables or new stories into numbers again
x_new = vectorizer.transform(new_story)

#making new predictions and geting the probability for each prediction
predictions = model.predict(x_new)
probabilities = model.predict_proba(x_new)
```

```python
#Making the actual prediction to see if the story is real or fake
for i, email in enumerate(new_story):
   result = "Real" if predictions[i] == 1 else "Fake"
   prob_real    = probabilities[i][1] * 100
   prob_fake = probabilities[i][0] * 100
```
