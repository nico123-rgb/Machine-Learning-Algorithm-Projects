```python
#importing all necessary libraries
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, association_rules
#https://www.kaggle.com/datasets/alyelbadry/market-basket-optimization?utm_source=chatgpt.com
```

```python
#Importing the dataset file
from google.colab import files
uploaded=files.upload()
df=pd.read_csv('Market_Basket_Optimisation.csv')
df.head()
```

```python
#This code ignores the null values
#We needed to use this code because we could set in list form because "floats" and "integers"
transactions = []

for i in range(len(df)):
    transactions.append([
        str(item) for item in df.iloc[i] if pd.notna(item)
    ])
```

```python
#Setting the items in list format in order to use the association rules algorithm
data = df.apply(lambda row: row.dropna().tolist(), axis=1).tolist()
data
```

```python
#Displaying the full table
df.head()
```

```python
#This cleans the data up in order to use the algorithm
#This sets them 0s and 1s
from mlxtend.preprocessing import TransactionEncoder
a=TransactionEncoder()
a_data=a.fit(data).transform(data)
df=pd.DataFrame(a_data, columns=a.columns_)
```

```python
#This replaces all 0s and 1s and true and false
df=df.replace(False,0)
df=df.replace(True,1)

df
```

```python
#This code set the minimum support and threshold fpr the association rules: support, confidence, lift, conviction.
#Also setting the antecedents and consequents
from mlxtend.frequent_patterns import apriori, association_rules

frequent_itemsets = apriori(df, min_support=0.01, use_colnames=True)
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.30)

rules[["antecedents","consequents","support","confidence","lift","conviction"]].head(10)
```
