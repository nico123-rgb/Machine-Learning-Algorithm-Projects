```python
df=pd.read_csv("ObesityDataSet_raw_and_data_sinthetic.csv")
df
#printing the dataset
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
label = le.fit_transform(df['family_history_with_overweight'])
label

#"family history with overweight" is turned from a string to 0 and 1
```

```python
df.drop("family_history_with_overweight", axis=1, inplace=True)
df["family_history_with_overweight"] = label
df
```

```python
df = df.drop(columns=['FAVC','FCVC','NCP','CAEC','SMOKE','CH2O','SCC','FAF','TUE','CALC','MTRANS','NObeyesdad','Gender'])
df

# removed the following columns as we dont need for our model
```

```python
old_cols=df.columns.values
new_cols=['family_history_with_overweight','Age','Height','Weight']
df=df.reindex(columns=new_cols)
df

#Rearanged the columns so that "famiily history with overweight" is first since we believe that this an important aspect of the dataset
```

```python
print(df.rename(columns={'family_history_with_overweight':'Family History Obesity','Height':'Height in Meters','Weight':'Weight in Kilograms'}))
df

#Renamed the columns so that the reader know what unit weight and height has
```

```python
#Done with Feature Engineering
```

```python
df
#Printing manipulated dataset
#Starting K-means clustering
```

```python
plt.scatter(df['Height'],df['Weight'],c='green',s=10)
plt.xlabel('Height')
plt.ylabel('Weight')

#plotting all the dataset points and labeling the x and y axis
```

```python
k=2
X=np.array(list(zip(df['Height'],df['Weight'])))
model=KMeans(n_clusters=k)
kmeans=model.fit(X)
labels=model.predict(X)
centroids=model.cluster_centers_
print(labels)
print(centroids)

#k is 2  as we only wanted two clusters
#called the array to use x and y
#the centoids is called and centered
```

```python
c=['blue','red','yellow','green']
colors=[c[i] for i in labels]

plt.scatter(df['Height'],df['Weight'],c=colors, s=10)
plt.scatter(centroids[:,0], centroids[:,1],marker='*', s=100, c='black')
plt.xlabel('Height')
plt.ylabel('Weight')

#seperated the points to two clusters and the centroids are placed
#colors and size are also assigned
```

```python
cluster=model.predict([[1.5,90]])[0]
print(c[cluster])

cluster=model.predict([[1.5,150]])[0]
print(c[cluster])

cluster=model.predict([[1.8,120]])[0]
print(c[cluster])

#predicting what cluster any given point will be a part of
```

```python
from sklearn.metrics import silhouette_score

model=KMeans(n_clusters=2)
model.fit(X)
label=model.predict(X)

print(f'silhouette Score (n=2): {silhouette_score(X,label)}')

#function that lets you know how many centroids are best for this dataset
```
